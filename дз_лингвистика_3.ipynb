```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
#!pip install python-docx
```

```python
from docx import Document

def extract_text_from_docx(file_path):
    doc = Document(file_path)
    full_text = []
    for paragraph in doc.paragraphs:
        full_text.append(paragraph.text)
    return '\n'.join(full_text)

text1 = extract_text_from_docx('библия.docx')
text2 = extract_text_from_docx('коран.docx')
text3 = extract_text_from_docx('типитака.docx')
text4 = extract_text_from_docx('тора.docx')

documents = [text1, text2, text3, text4]
doc_names = ["Библия", "Коран", "Типитака", "Тора"]
```

```python
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)
similarity_matrix = cosine_similarity(tfidf_matrix)
```

```python
print("Матрица косинусного сходства:\n")
for i in range(len(doc_names)):
    for j in range(i + 1, len(doc_names)):
        similarity_score = similarity_matrix[i][j]
        print(f"{doc_names[i]} и {doc_names[j]} : сходство = {similarity_score:.4f}")
```

удивительно, что Коран и Типитака похожи друг на друга больше, чем Библия и Тора) даже Библия и Типитака похожи больше чем Библия и Тора. думаю, это получилось из-за того, что я взяла во всех книгах рандомные главы
